# 00 — References

Consolidated bibliography for the LunarLander baseline pipeline (Project 50.1, MSc Robotics & AI, University of Aberdeen). Every cited claim in notebooks 01–07 traces back to one of these references. Lift this list directly into the dissertation Bibliography section.

## Reinforcement-learning algorithms

**PPO — Proximal Policy Optimization.**  Schulman, J., Wolski, F., Dhariwal, P., Radford, A., & Klimov, O. (2017). *Proximal Policy Optimization Algorithms.* arXiv:1707.06347. — The clipped surrogate objective $\mathcal{L}^{CLIP}$ used in notebooks 03 and 05.

**SAC — Soft Actor-Critic.**  Haarnoja, T., Zhou, A., Abbeel, P., & Levine, S. (2018). *Soft Actor-Critic: Off-Policy Maximum Entropy Deep Reinforcement Learning with a Stochastic Actor.* ICML 2018. — Maximum-entropy continuous-control baseline used in notebook 04.

**TD3 / twin-critic device.**  Fujimoto, S., van Hoof, H., & Meger, D. (2018). *Addressing Function Approximation Error in Actor-Critic Methods.* ICML 2018. — Source of the twin Q-network construction inherited by SAC.

**GAE — Generalised Advantage Estimation.**  Schulman, J., Moritz, P., Levine, S., Jordan, M., & Abbeel, P. (2016). *High-Dimensional Continuous Control Using Generalized Advantage Estimation.* ICLR 2016. — The $\hat{A}_t$ estimator with $\lambda = 0.98$ used in the PPO configs.

## Robustness, domain randomisation, and noisy-actuator analogues

**Domain Randomisation.**  Tobin, J., Fong, R., Ray, A., Schneider, J., Zaremba, W., & Abbeel, P. (2017). *Domain Randomization for Transferring Deep Neural Networks from Simulation to the Real World.* IROS 2017. — Justifies the PPO+DR baseline as the *training-time* compensator for environment variability.

**Sim-to-real with adaptive DR.**  Peng, X. B., Andrychowicz, M., Zaremba, W., & Abbeel, P. (2018). *Sim-to-Real Transfer of Robotic Control with Dynamics Randomization.* ICRA 2018. — Discussion of how DR ranges relate to the deployment-time fault distribution.

**NISQ gate drift (Quantum Bridge motivation).**  Preskill, J. (2018). *Quantum Computing in the NISQ era and beyond.* Quantum 2, 79. — Source for the structural analogy between non-stationary actuator noise and quantum gate drift used in the thesis introduction.

**Actuator fault detection in robotics.**  Verma, V., Gordon, G., Simmons, R., & Thrun, S. (2004). *Real-time Fault Diagnosis.* IEEE Robotics & Automation Magazine, 11(2):56-66. — Classical reference for the prediction-error / model-residual school of fault diagnosis.

## Evaluation methodology

**Multi-seed RL evaluation.**  Henderson, P., Islam, R., Bachman, P., Pineau, J., Precup, D., & Meger, D. (2018). *Deep Reinforcement Learning that Matters.* AAAI 2018. — Source for the n=5 seed convention, the requirement to report per-seed variance, and the warning that single-seed RL results are misleading.

**Cliff's δ effect size.**  Romano, J., Kromrey, J. D., Coraggio, J., & Skowronek, J. (2006). *Appropriate statistics for ordinal level data: Should we really be using t-test and Cohen's d for evaluating group differences on the NSSE and other surveys?* Annual Meeting of the Florida Association of Institutional Research. — Magnitude thresholds (negligible / small / medium / large) used in `src/utils/stats.py`.

**False discovery rate.**  Benjamini, Y., & Hochberg, Y. (1995). *Controlling the False Discovery Rate: a Practical and Powerful Approach to Multiple Testing.* JRSS-B, 57(1):289-300. — BH correction applied to pairwise comparisons in `pairwise_stats_table`.

**Welch's t-test.**  Welch, B. L. (1947). *The generalization of "Student's" problem when several different population variances are involved.* Biometrika, 34(1/2):28-35. — The unequal-variance two-sample test used because per-algorithm variances differ (e.g. PPO σ ≈ 9 vs PPO+DR σ ≈ 18).

**Mann-Whitney U.**  Mann, H. B., & Whitney, D. R. (1947). *On a test of whether one of two random variables is stochastically larger than the other.* Annals of Mathematical Statistics, 18(1):50-60. — Non-parametric companion to Welch's t for distribution-shift detection.

## Tabular value-based RL and change-point detection

**Q-learning (tabular).**  Watkins, C. J. C. H., & Dayan, P. (1992). *Q-learning.* Machine Learning, 8(3-4):279-292. — The off-policy temporal-difference update rule used in notebook 09 as the simplest possible value-based baseline.

**Action-space discretisation for tabular RL.**  Sutton, R. S., & Barto, A. G. (2018). *Reinforcement Learning: An Introduction* (2nd ed.), Chapter 6. MIT Press. — Standard reference for state-space binning strategies; used in notebook 09 to map LunarLander's 8-D continuous observation into a tractable lookup table.

**Change-point detection — PELT.**  Killick, R., Fearnhead, P., & Eckley, I. A. (2012). *Optimal Detection of Changepoints With a Linear Computational Cost.* JASA, 107(500):1590-1598. — One of the change-point detection routines applied in notebook 08 to internal-signal time series.

**Change-point detection survey.**  Aminikhanghahi, S., & Cook, D. J. (2017). *A Survey of Methods for Time Series Change Point Detection.* Knowledge and Information Systems, 51(2):339-367. — Background reference for the sliding-window-z and CUSUM detectors used alongside PELT in notebook 08.

**Sliding-window z-score baseline.**  Page, E. S. (1954). *Continuous Inspection Schemes.* Biometrika, 41(1/2):100-115. — Original CUSUM-style change-point detection; the practical lineage of the rolling-statistic detectors in notebook 08.

## Related work additions (NB16-NB19) --- the publication-gap references

These close the reviewer-facing gap identified in the paper review: the study previously cited change-point statistics and RL algorithms but no *fault detection and isolation* (FDI), no RL anomaly-detection work, and no quantum-drift sources for the NISQ analogy.

### Classical fault detection and isolation (FDI) --- NB16, NB18
- Isermann, R. (2006). *Fault-Diagnosis Systems: An Introduction from Fault Detection to Fault Tolerance.* Springer.  [The standard FDI reference: detection/isolation/identification pipeline; analytical redundancy. NB18's forward model IS analytical redundancy; NB16's attribution IS isolation.]
- Gertler, J. (1998). *Fault Detection and Diagnosis in Engineering Systems.* Marcel Dekker.  [Structured residuals --- the design principle that NB16's action-conditional residual ratio recovers for free.]
- Willsky, A. S. (1976). A survey of design methods for failure detection in dynamic systems. *Automatica*, 12(6), 601-611.  [The founding survey; cite once in Related Work.]
- Basseville, M., & Nikiforov, I. V. (1993). *Detection of Abrupt Changes: Theory and Application.* Prentice Hall.  [The SPC canon behind CUSUM/Page-Hinkley; grounds NB19's detector comparison.]
- Roberts, S. W. (1959). Control chart tests based on geometric moving averages. *Technometrics*, 1(3), 239-250.  [EWMA control chart --- NB19.]

### Anomaly / novelty detection in RL --- NB16, NB18
- Sedlmeier, A., Gabor, T., Phan, T., Belzner, L., & Linnhoff-Popien, C. (2020). Uncertainty-based out-of-distribution classification in deep reinforcement learning. *ICAART*.  [OOD detection from policy uncertainty --- the nearest neighbour to the entropy-signal finding; NB16's H-state vs H-conflict decomposition separates our result from theirs.]
- Haider, T., Roza, F. S., Eilers, D., Roscher, K., & Guennemann, S. (2023). Out-of-distribution detection for reinforcement learning agents in the real world (workshop line of work).  [Verify exact venue/year during final reference pass.]
- Chalapathy, R., & Chawla, S. (2019). Deep learning for anomaly detection: A survey. arXiv:1901.03407.  [One-line framing citation.]
- Muller, R., Illium, S., Phan, T., Haider, T., & Linnhoff-Popien, C. (2022). Towards anomaly detection in reinforcement learning. *AAMAS (Blue Sky Ideas)*.  [Positions anomaly detection as an open RL problem --- supports the gap claim.]

### Fault-tolerant / fault-adaptive RL --- Related Work paragraph 2
- Ahmed, I., Quinones-Grueiro, M., & Biswas, G. (2020). Complementary meta-reinforcement learning for fault-adaptive control. *PHM Society Conference*.  [Adaptation WITHOUT detection --- the complement of this study's detection-without-adaptation; the two papers bracket NB17's detect-then-recover.]
- Nagabandi, A., Clavera, I., et al. (2019). Learning to adapt in dynamic, real-world environments through meta-reinforcement learning. *ICLR*.  [Meta-RL online adaptation; the 'SAC compensates implicitly' result is the zero-shot version.]
- Fisac, J. F., et al. (2019). A general safety framework for learning-based control systems. *IEEE TAC.*  [Safe-mode switching context for NB17's recovery architecture.]
